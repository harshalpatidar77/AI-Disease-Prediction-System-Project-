```python
# imports and paths 
import os 
import numpy as np 
import pandas as pd 
np.random.seed(42) 
RAW_DIR = "../data/raw" 
PROCESSED_DIR = "../data/raw" 
os.makedirs(RAW_DIR, exist_ok=True)
os.makedirs(PROCESSED_DIR, exist_ok=True) 
```

```python
# Generate synthetic dataset  
n = 1500

diseases = [
    "Flu",
    "Pneumonia",
    "Diabetes",
    "Hypertension",
    "Migraine"
]

disease = np.random.choice(
    diseases,
    size=n,
    p=[0.22, 0.18, 0.20, 0.20, 0.20]
)

df = pd.DataFrame({
    "age": np.random.randint(18, 80, n),
    "gender": np.random.choice(["Male", "Female"], n),
    "temperature": np.round(np.random.normal(98.6, 1.8, n), 1),
    "heart_rate": np.random.randint(55, 125, n),
    "blood_pressure": np.random.randint(90, 180, n),
    "glucose": np.random.randint(70, 220, n),
    "cholesterol": np.random.randint(130, 280, n),

    "fever": np.random.choice(["Yes", "No"], n),
    "cough": np.random.choice(["Yes", "No"], n),
    "fatigue": np.random.choice(["Yes", "No"], n),
    "headache": np.random.choice(["Yes", "No"], n),
    "chest_pain": np.random.choice(["Yes", "No"], n),
    "breathing_difficulty": np.random.choice(["Yes", "No"], n),
    "nausea": np.random.choice(["Yes", "No"], n),
    "disease": disease
}) 
# Add realistic disease-related patterns
df.loc[df["disease"] == "Flu", "fever"] = np.random.choice(
    ["Yes", "No"],
    size=(df["disease"] == "Flu").sum(),
    p=[0.85, 0.15]
)

df.loc[df["disease"] == "Flu", "cough"] = np.random.choice(
    ["Yes", "No"],
    size=(df["disease"] == "Flu").sum(),
    p=[0.75, 0.25]
)

df.loc[df["disease"] == "Pneumonia", "breathing_difficulty"] = np.random.choice(
    ["Yes", "No"],
    size=(df["disease"] == "Pneumonia").sum(),
    p=[0.70, 0.30]
)

df.loc[df["disease"] == "Pneumonia", "cough"] = np.random.choice(
    ["Yes", "No"],
    size=(df["disease"] == "Pneumonia").sum(),
    p=[0.85, 0.15]
)

df.loc[df["disease"] == "Diabetes", "glucose"] = np.random.randint(
    130, 220,
    (df["disease"] == "Diabetes").sum()
)

df.loc[df["disease"] == "Hypertension", "blood_pressure"] = np.random.randint(
    140, 180,
    (df["disease"] == "Hypertension").sum()
)

df.loc[df["disease"] == "Migraine", "headache"] = np.random.choice(
    ["Yes", "No"],
    size=(df["disease"] == "Migraine").sum(),
    p=[0.90, 0.10]
)

df.head()
```

```python
# Add missing values 
missing_columns = [
    "age",
    "temperature",
    "glucose",
    "blood_pressure",
    "fatigue",
    "headache"
]

for column in missing_columns:
    indices = np.random.choice(
        df.index,
        size=int(0.03 * len(df)),
        replace=False
    )
    df.loc[indices, column] = np.nan

print("Missing values added.") 
```

```python
# Save dataset 
dataset_path = os.path.join(
    RAW_DIR,
    "disease_dataset.csv"
)

df.to_csv(dataset_path, index=False)

print(f"Dataset saved to: {dataset_path}")
print("Shape:", df.shape) 
```
