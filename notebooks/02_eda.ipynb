```python
# imports 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
df = pd.read_csv("../data/raw/disease_dataset.csv") 
df.head() 
```

```python
# Dataset information 
print("Shape:", df.shape) 
print("\nInformation:") 
df.info() 
print("\nStatistics:")
display(df.describe()) 
```

```python
# Missing values 
missing = df.isnull().sum() 
print(missing[missing > 0]) 
```

```python
# Disease distribution 
plt.figure(figsize=(8, 5)) 
sns.countplot(
    data = df, 
    x = "disease" 
) 
plt.title("Disease Distribution") 
plt.xticks(rotation=30)
plt.tight_layout() 
plt.savefig("../reports/disease_distribution.png")
plt.show() 
```

```python
# Age distribution 
plt.figure(figsize=(8, 5)) 
sns.histplot(
    data = df, 
    x = "age", 
    kde=True 
) 
plt.title("Age Distribution")
plt.tight_layout() 
plt.show() 
```

```python
# numeric correlation 
numeric_df = df.select_dtypes(include=np.number) 
plt.figure(figsize=(10, 7)) 
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
) 
plt.title("Numerical Feature Correlation")
plt.tight_layout() 
plt.show() 
```

```python
# Disease vs glucose 
plt.figure(figsize=(10, 6)) 
sns.boxplot(
    data=df,
    x="disease",
    y="glucose"
) 
plt.title("Glucose Distribution by Disease")
plt.xticks(rotation=30)
plt.tight_layout() 
plt.show()  
```
